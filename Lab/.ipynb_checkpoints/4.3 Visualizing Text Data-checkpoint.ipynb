```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
plt.style.use('fivethirtyeight')
```

# The Dataset - Harry Potter Reviews

This dataset is a collection of Harry Potter movie reviews from Kaggle, including user ratings, comments, favorite characters, and basic demographic info.

```python
df = pd.read_csv('harry_potter_reviews.csv')
```

```python
df.head()
```

# Visualizations

# 1. Favorite Character Bar Chart

```python
char_counts = df["favourite_character"].value_counts().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
char_counts.plot(kind="bar")
plt.title("Favorite Harry Potter Characters in Reviews")
plt.xlabel("Character")
plt.ylabel("Number of Mentions")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

This bar chart shows which Harry Potter characters people picked as their favorites in the reviews. You can see Ron, Hermione, and Harry at the top, which makes sense since they’re the main trio and appear in almost every major scene.

# 2. Average Rating by Year Line Chart

```python
df["date"] = pd.to_datetime(df["date"])

ratings_by_year = df.groupby(df["date"].dt.year)["rating"].mean()

plt.figure(figsize=(8, 5))
ratings_by_year.plot(marker="o")
plt.title("Average Review Rating by Year")
plt.xlabel("Year")
plt.ylabel("Average Rating")
plt.xticks(ratings_by_year.index)
plt.ylim(0, 5) 
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

This line chart shows how the average review rating changed from 2002 to 2005. The ratings stay pretty steady overall, with a small dip in 2004, which could mean that year’s movie didn’t hit quite as hard compared to the others.

# 3. Average Rating by Country Bar Chart

```python
country_ratings = df.groupby("user_country")["rating"].mean().sort_values(ascending=False)

plt.figure(figsize=(12, 6))
country_ratings.plot(kind="bar")
plt.title("Average Review Rating by Country")
plt.xlabel("Country")
plt.ylabel("Average Rating")
plt.xticks(rotation=45, ha="right")
plt.ylim(0, 5)  
plt.tight_layout()
plt.show()
```

This bar chart compares the average movie rating by country. Some countries rated the movie slightly higher than others, but overall most countries gave it around 4 stars, showing pretty consistent global appreciation.
